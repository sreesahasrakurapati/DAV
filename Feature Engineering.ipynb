```python

# Step 1: Import required libraries

import pandas as pd
import numpy as np

# Create a sample dataset
data = {
    'Study_Hours': [2, 4, 6, 8, 10, 3, 5, 7, 9, 12],
    'Attendance': [60, 65, 70, 75, 80, 62, 68, 78, 85, 95],
    'Assignment_Score': [40, 50, 55, 65, 70, 45, 60, 72, 80, 90],
    'Exam_Score': [45, 50, 60, 65, 75, 48, 58, 70, 82, 95]
}
```

```python

# Step 2: Convert data into a DataFrame

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

print("\nStatistical Summary:")
print(df.describe())
```

```python

# Step 3: Import MinMaxScaler

from sklearn.preprocessing import MinMaxScaler

# Initialize MinMaxScaler
scaler = MinMaxScaler()

# Apply normalization
normalized_data = scaler.fit_transform(df)

# Convert the result into a DataFrame
normalized_df = pd.DataFrame(
    normalized_data,
    columns=df.columns
)

print("Normalized Dataset:")
print(normalized_df.round(2))

```

```python

# Step 4: Check minimum values

print("Minimum values:")
print(normalized_df.min())

# Check maximum values
print("\nMaximum values:")
print(normalized_df.max())
```

```python

# Step 5: Import StandardScaler

from sklearn.preprocessing import StandardScaler

# Initialize StandardScaler
scaler = StandardScaler()

# Apply standardization
standardized_data = scaler.fit_transform(df)

# Convert the result into a DataFrame
standardized_df = pd.DataFrame(
    standardized_data,
    columns=df.columns
)

print("Standardized Dataset:")
print(standardized_df.round(2))
```

```python

# Step 6: Check the mean of each feature

print("Mean of each feature:")
print(standardized_df.mean().round(2))

# Check standard deviation
print("\nStandard deviation of each feature:")
print(standardized_df.std(ddof=0).round(2))
```

```python

# Step 7: Import PCA and Matplotlib

from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
```

```python

# Step 8: Standardize the original dataset

scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)
```

```python

# Step 9: Apply PCA with two principal components

pca = PCA(n_components=2)

pca_data = pca.fit_transform(scaled_data)

# Create a DataFrame for PCA results
pca_df = pd.DataFrame(
    pca_data,
    columns=['PC1', 'PC2']
)

print("PCA Transformed Data:")
print(pca_df.round(2))
```

```python

# Step 10: Display explained variance ratio

print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)

# Display total variance retained
print("\nTotal Variance Retained:")
print(pca.explained_variance_ratio_.sum())
```
